```python
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root: TreeNode) -> int:
    seen_nodes = set()
    duplicate_found = [False]

    def traverse(node):
        if not node or duplicate_found[0]:
            return
        if id(node) in seen_nodes:
            duplicate_found[0] = True
            return
        seen_nodes.add(id(node))
        traverse(node.left)
        traverse(node.right)

    traverse(root)
    return 1 if duplicate_found[0] else 0
```

```python
# Your answer here
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    length_nums = len(nums)
    i = 0

    while i < length_nums:
        if nums[i] == 0:
            nums.pop(i)
            nums.append(0)
            length_nums -= 1
        else:
            i += 1
    return nums
   
```

# Your answer here
This function loops through the list from the beginning. If it encounters a zero, it removes it (pop(i)), and appends a zero to the end of the list (append(0)). This effectively moves the current zero to the end while keeping non-zero elements in the same relative order.

The length_nums variable is decreased whenever a zero is found so the loop avoids revisiting the newly added zero at the end. The code returns the modified list once all elements are processed.



-   Explain the problem’s time and space complexity in your own words.


# Your answer here
Time Complexity:
The code uses a while loop that may run up to n times, but inside the loop, the pop(i) operation takes O(n) time in the worst case (since all elements after i must be shifted). As a result, this makes the overall time complexity O(n²) in the worst case.

Space Complexity:
The list is modified in-place, and no new lists or data structures are created. So, space complexity is O(1).
